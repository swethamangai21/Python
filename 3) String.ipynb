```python
# Two Pointers- uses two variables (pointers) to access different positions in a string

# Instead of using nested loops, we use two pointers to solve certain problems more efficiently

'''
There are two common approaches:
1. Opposite Direction: One pointer starts at the beginning, and another starts at the end.
2. Same Direction: Both pointers move forward, often at different speeds.
'''

'''
s = "madam"

Index:      0   1   2   3   4
            m   a   d   a   m
            ↑               ↑
           left            right

initially, 
    left=0
    right=len(s)-1

Step 1:
m == m 

Step 2:
a == a 

Step 3:
Both pointers reach the middle.

Since all corresponding characters match, the string is a palindrome.

After every successful comparison:
left += 1
right -= 1
'''

'''
check palindrome- Compare characters from both ends
reverse a string- Swap characters from opposite ends
valid palindromw ignoring symbols- Skip unwanted characters
compare characters symmetrically- Process both ends simultaneously
reverse only vowels- Find vowels from opposite directions
'''
```

```python
# Check whether a string is a palindrome

s='madam'

left=0
right=len(s)-1

isPalindrome=True

while left<right:
    if s[left]!=s[right]:
        isPalindrome=False

    left+=1
    right-=1

print(isPalindrome)
```

```python
# Reverse a string

s = ["h", "e", "l", "l", "o"]

left=0
right=len(s)-1

while left<right:
    s[left], s[right]= s[right], s[left]

    left+=1
    right-=1

print(s)
```

```python
# Sliding Window- a technique used to process a continuous portion of a string or array, called a window

# Instead of repeatedly processing the same elements, we move the window forward and reuse the information we've already calculated

'''
We need to examine every substring of length 3:

String: a b c d e f

Window 1: [a b c] d e f
Window 2:  a [b c d] e f
Window 3:  a b [c d e] f
Window 4:  a b c [d e f]

Notice:
- Window size always remains 3.
- The window moves one position at a time.
- One character leaves, and another enters.
This is the fundamental idea behind Sliding Window.
'''

'''
Substring of length K- Substring of length K
Maximum/minimum in K elements- Evaluate every fixed-size window
Count characters in each substring of length K- Update counts as the window moves
Consecutive elements- Windows cover contiguous portions
Average of K elements- Reuse the previous window's sum
'''
```

```python
# Maximum Vowels in a Substring of Length K

s = "abciiidef"
k = 3

count=0

# step 1: count vowels in the first window
for i in range(k):
    if s[i] in 'aeiou':
        count+=1

max_count=count

# step 2: slide the window
# s[i-k] --> leaving element
# s[i] --> entering element
for i in range(k, len(s)):             
    if s[i-k] in 'aeiou':
        count-=1
    if s[i] in 'aeiou':
        count+=1

    max_count=max(max_count, count)

print(max_count)
```

```python
# Maximum Sum of K Consecutive Elements

nums = [2, 1, 5, 1, 3, 2]
k = 3

sum=0

for i in range(k):
    sum=sum+nums[i]

max_sum=sum

for i in range(k, len(nums)):
    sum=sum-nums[i-k]
    sum=sum+nums[i]

    max_sum=max(max_sum, sum)

print(max_sum)
```

```python
# Stack- a data structure that follows the LIFO principle (Last In, First Out)

# In Python, we can use a list as a stack

'''
-Valid parentheses: "({[]})"
-Remove adjacent duplicates
-Balanced brackets
-Undo operations
-Next greater element
'''
```

```python
# Remove Adjacent Duplicates

# if stack: --> executes when the stack is not empty
# if not stack: --> executes when the stack is empty

s = "abbacb"

stack=[]

for i in s:
    if stack and stack[-1]==i:
        stack.pop()
    else:
        stack.append(i)
    

print(stack)
```

```python
# Sorting + Two Pointers

'''
Find a pair with a target sum- adjust pointers based on the sum
Find triplets with a target sum- fix one element and use two pointers
Find pairs with a given difference- use sorted order to guide pointer movement
Find closest pair sum- compare sums while moving pointers
'''
```

```python
# Two Sum Using Two Pointers

nums = [2, 7, 11, 15]
target = 9

# step 1: sort the array
nums.sort()

# step 2: initialize pointers
left=0
right=len(nums)-1

# step 3: calculate sum
'''
2 + 15 = 17

Our target is 9.

Since 17 > 9, we need a smaller sum.

Move the right pointer backward.
2   7   11   15
↑       ↑
L       R
'''

while left<right:
    sum=nums[left]+nums[right]

    if sum==target:
        print(nums[left], nums[right])
        break

    elif sum>target:
        right-=1

    else:
        left+=1

```

```python
# check whether the sum of two numbers has the target

nums = [2, 7, 11, 15]
target = 9

nums.sort()

left = 0
right = len(nums) - 1

found = False

while left < right:

    current_sum = nums[left] + nums[right]

    if current_sum == target:
        found = True
        break

    elif current_sum < target:
        left += 1

    else:
        right -= 1

print(found)
```
