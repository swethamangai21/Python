```python
'''Q1. Extract Numbers

From a string extract all numbers.

Example

text = "Order123 price45 quantity9"

Expected

['123','45','9']'''

import re
text = "Order123 price45 quantity9"
print(re.findall(r'\d+',text))
```

```python
'''Q2. Extract Words Only

Extract only alphabet words.

Example

text = "hello123 world_45 python!"

Expected

['hello','world','python']'''

import re
text = "hello123 world_45 python!"
print(re.findall('[a-zA-Z]+',text))
```

```python
'''Q3. Find All Email Addresses

Example

text = "Contact us at test@gmail.com or admin@yahoo.com"

Expected

['test@gmail.com','admin@yahoo.com']'''

import re
text = "Contact us at test@gmail.com or admin@yahoo.com"
print(re.findall('\w+@\w+\.\w+',text))
```

```python
'''Q4. Extract All 4 Digit Numbers

Example

text = "Years 1999 2020 123 45678"

Expected

['1999','2020']'''

import re
text = "Years 1999 2020 123 45678"
print(re.findall(r'\b\d{4}\b',text))
```

```python
'''Q5. Find Words Starting With Capital Letter

Example

text = "Alice and Bob live in New York"

Expected

['Alice','Bob','New','York']'''

import re
text = "Alice and Bob live in New York"
print(re.findall('[A-Z][a-z]+',text))
```

```python
'''Q6. Extract Phone Numbers (10 digits)

Example

text = "Call 9876543210 or 9123456780"

Expected

['9876543210','9123456780']'''

import re
text = "Call 9876543210 or 9123456780"
print(re.findall(r'\d{10}',text))
```

```python
'''Q7. Extract Words With Exactly 5 Letters

Example

text = "apple mango car house table dog tiger"

Expected

['apple','mango','house','tiger']'''

import re
text = "apple mango car house table dog tiger"
print(re.findall(r'\b[a-zA-Z]{5}\b',text))
```

```python
'''Q8. Find Words Ending With "ing"

Example

text = "I am coding and learning regex while eating"

Expected

['coding','learning','eating']'''

import re
text = "I am coding and learning regex while eating"
print(re.findall(r'\b[a-zA-Z]+ing\b',text))
```

```python
'''Q9. Replace All Digits With X

Example

text = "My number is 98765"

Expected

"My number is XXXXX"'''

import re
text = "My number is 98765"
print(re.sub('\d','X',text))
```

```python
'''Q10. Split String By Multiple Spaces

Example

text = "Python   SQL   Excel"

Expected

['Python','SQL','Excel']'''

import re
text = "Python   SQL   Excel"
print(re.split(r"\s+",text))
```

```python
'''Q11. Extract Dates

Example

text = "Today is 16-03-2026 and tomorrow is 17-03-2026"

Expected

['16-03-2026','17-03-2026']'''

import re
text = "Today is 16-03-2026 and tomorrow is 17-03-2026"
print(re.findall(r'\d{2}-\d{2}-\d{4}',text))
```

```python
'''Q12. Extract Hashtags

Example

text = "Learning #Python and #DataScience today"

Expected

['#Python','#DataScience']'''

import re
text = "Learning #Python and #DataScience today"
print(re.findall(r'#[a-zA-Z]+',text))
```

```python
'''Q13. Extract Words With Only Vowels

Example

text = "aeiou sky area idea queue"

Expected

['aeiou']'''

import re
text = "aeiou sky area idea queue"
print(re.findall(r'\b[aeiou]+\b',text))
```

```python
'''Q14. Extract Words With Both Letters and Numbers

Example

text = "abc123 hello test45 python"

Expected

['abc123','test45']'''

import re
text = "abc123 hello test45 python"
print(re.findall(r'\b(?=\w*[a-zA-Z])(?=\w*\d)\w+\b',text))
```

```python
'''Q15. Remove All Special Characters

Example

text = "Hello!@# Python$%^ 123"

Expected

"Hello Python 123"'''

import re
text = "Hello!@# Python$%^ 123"
print(re.sub('\W+',' ',text))
```

```python
'''Q16. Extract Currency Values

Example

text = "Price is $45 and discount is $5"

Expected

['$45','$5']'''

import re
text = "Price is $45 and discount is $5"
print(re.findall('[$]\d+',text))
```

```python
'''Q17. Extract Duplicate Words

Example

text = "this is is a test test"

Expected

['is','test']'''

import re
text = "this is is a test test"
pattern=r'\b(\w+)\s+\1\b'
print(re.findall(pattern,text))

```

```python
'''Q18. Extract Words With At Least 2 Consecutive Vowels

Example

text = "book good sky queue moon"

Expected

['book','good','queue','moon']'''

import re
text = "book good sky queue moon"
print(re.findall('[a-zA-Z]*[aeiou][aeiou][a-zA-Z]*',text))
```

```python
'''Q19. Validate Strong Password

Conditions

at least 8 characters

1 uppercase

1 lowercase

1 digit

Example

password = "Python123"

Expected

True'''

import re
password = "Python123"
pattern=r'\b(?=.*[a-z])(?=.*[A-Z])(?=.*\d).{8,}\b'
if re.match(pattern,password):
    print(True)
else:
    print(False)
```

```python
'''Q20. Extract IP Address

Example

text = "Server IP is 192.168.1.1 and backup is 10.0.0.1"

Expected

['192.168.1.1','10.0.0.1']'''

import re
text = "Server IP is 192.168.1.1 and backup is 10.0.0.1"
print(re.findall(r'\b\d+\.\d+\.\d+\.\d+\b',text))
```
