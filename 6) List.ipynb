```python
'''Question 4: Count Even and Odd Numbers

Problem:
Given a list of numbers, count how many are even and how many are odd.

Example:
Input:

[1, 2, 3, 4, 5, 6]

Output:

Even: 3
Odd: 3'''

l=[1, 2, 3, 4, 5, 6]
ecount=0
ocount=0
for i in l:
    if i%2==0:
        ecount+=1
    else:
        ocount+=1
print('Even:',ecount)
print('Odd:',ocount)
```

```python
'''Question 9: Move All Zeros to End

Problem:
Given a list, move all zeros to the end while keeping the order of non-zero elements.

Example:
Input:

[0, 1, 0, 3, 12]

Output:

[1, 3, 12, 0, 0]'''

l=[0, 1, 0, 3, 12]
for i in l:
    if i==0:
        l.remove(i)
        l.append(0)
print(l)
```

```python
'''Question 10: Find Frequency of Each Element

Problem:
Given a list, count how many times each element appears.

Example:
Input:

[1, 2, 2, 3, 3, 3]

Output:

1 : 1
2 : 2
3 : 3'''

l=[1, 2, 2, 3, 3, 3]
for i in range(len(l)):
    if l[i] not in l[:i]:
        print(l[i],':',l.count(l[i]))
```

```python
'''Question 11: Check if List is Palindrome

Problem:
A list is a palindrome if it reads the same forwards and backwards.
Check if a given list is a palindrome.

Example:
Input:

[1, 2, 3, 2, 1]

Output:

Palindrome'''

l=[1, 2, 3, 2, 1]
if l==l[::-1]:
    print('Palindrome')
else: 
    print('Not Palindrome')
```

```python
'''Question 19: Flatten a Nested List

Problem:
Given a list that contains other lists inside it, convert it into a single flat list.

Example:
Input:

nested = [1, [2, 3], [4, [5, 6]]]

Output:

[1, 2, 3, 4, 5, 6]'''

nested = [1, [2, 3], [4, [5, 6]]]
flat = []

for i in nested:              
    if isinstance(i, list):   
        for j in i:           
            if isinstance(j, list):   
                for k in j:           
                    flat.append(k)    
            else:
                flat.append(j)       
    else:
        flat.append(i)                

print(flat)

print('*'*10)

def flatten(nested):
    flat=[]
    for i in nested:
        if isinstance(i,list):
            flat.extend(flatten(i))
        else:
            flat.append(i)
    return flat
nested = [1, [2, 3], [4, [5, 6]]]
print(flatten(nested))


print('*'*10)
nested = [1, [5, 2], [3, [9, 4]]]

flat = nested[:]

while any(isinstance(i, list) for i in flat):
    new = []
    for i in flat:
        if isinstance(i, list):
            new.extend(i)
        else:
            new.append(i)
    flat = new

print(max(flat))
```

```python
'''Question 20: Find Sublist in List

Problem:
Given a list l and a sublist sub, check if sub exists in l in the same order.

Example:
Input:

l = [1, 2, 3, 4, 5]
sub = [2, 3, 4]

Output:

True'''

l = [1, 2, 3, 4, 5]
sub = [2, 3, 4]

found = False
for i in range(len(l) - len(sub) + 1):
    if l[i:i+len(sub)] == sub:
        found = True
        break

print(found)
```
