```python
'''1. Count Character Frequency

Given a string, count the frequency of each character without using Counter.

Input: "programming"

Output: {'p': 1, 'r': 2, 'o': 1, 'g': 2, 'a': 1, 'm': 2, 'i': 1, 'n': 1}'''

s="programming"
for i in range(len(s)):
    if s[i] not in s[:i]:
        print(s[i],':',s.count(s[i]),end=', ')

freq={}
for i in s:
    freq[i]=freq.get(i,0)+1
print(freq)
```

```python
'''Question 2: Most Frequent Element in a List

Problem:
Given a list of numbers, find the most frequent element using a dictionary without using collections.Counter.

Example:
Input: [3, 5, 2, 3, 7, 2, 3]
Output: Most frequent element is 3, frequency: 3 '''

l=[3, 5, 2, 3, 7, 2, 3]
freq={}
for i in l:
    freq[i]=freq.get(i,0)+1
print(freq)

value=0
key=None
for k,v in freq.items():
    if v>value:
        value=v
        key=k
print(key,':',value)
```

```python
'''Question 3: Replace Most Frequent Element with 0

Problem:
Given a list of numbers, find the most frequent element and replace all its occurrences with 0 using a dictionary.

Example:
Input: [3, 5, 2, 3, 7, 2, 3]
Output: [0, 5, 2, 0, 7, 2, 0]'''

l=[3, 5, 2, 3, 7, 2, 3]
freq={}
for i in l:
    freq[i]=freq.get(i,0)+1
key=None
value=0
for k,v in freq.items():
    if v>value:
        value=v
        key=k
for i in range(len(l)):
    if l[i]==key:
        l[i]=0
print(l)
```

```python
'''Question 4: Group Elements by Frequency

Problem:
Given a list of elements, group them into a dictionary where keys are the elements and values are lists of all occurrences.

Example:
Input: [3, 3, 5, 2, 2, 2]
Output: {3: [3, 3], 5: [5], 2: [2, 2, 2]}'''

l=[3, 3, 5, 2, 2, 2]
freq={}
for i in l:
    freq[i]=freq.get(i,[])+[i]
print(freq)
```

```python
'''Question 6: Merge Dictionaries with Sum of Values

Problem:
Given two dictionaries, merge them. If a key appears in both, sum their values.

Example:

d1 = {'a': 2, 'b': 3}
d2 = {'b': 4, 'c': 5}

Output:

{'a': 2, 'b': 7, 'c': 5}'''

d1 = {'a': 2, 'b': 3}
d2 = {'b': 4, 'c': 5}
merged=d1.copy()
for k,v in d2.items():
    merged[k]=merged.get(k,0)+v
print(merged)
```

```python
'''Question 9: Count Pairs with Given Sum

Problem:
Given a list of integers and a target sum, count how many pairs of elements add up to the target using a dictionary.

Example:

arr = [1, 5, 7, -1, 5]
target = 6

Output:

Number of pairs = 3

Explanation:
Pairs are (1,5), (7,-1), (1,5) → total 3 pairs'''

arr = [1, 5, 7, -1, 5]
target = 6
res=[]
for i in range(len(arr)):
    for j in range(i + 1,len(arr)): 
            if arr[i]+arr[j]==target:
                res.append((arr[i],arr[j]))

print(res)
print(len(res))
```

```python
'''Question 12: Group Words by Length or First Letter Using Dictionary

Problem:
Given a list of words, group them by their length (or first letter) using a dictionary.

Example 1 (group by length):

words = ["cat", "dog", "apple", "bat", "ant"]

Output:

{3: ["cat", "dog", "bat", "ant"], 5: ["apple"]}'''

words = ["cat", "dog", "apple", "bat", "ant"]
grouped={}
for i in words:
    key=len(i)
    grouped.setdefault(key,[]).append(i)
print(grouped)
```

```python
'''Question 14: Merge Dictionaries, Keep Max Value

Problem:
You are given two dictionaries. Merge them into one.

If a key appears in both dictionaries, keep the maximum value.

Otherwise, keep the value as is.

Example:

d1 = {'a': 10, 'b': 20}
d2 = {'a': 15, 'c': 5}

Output: {'a': 15, 'b': 20, 'c': 5}'''

d1 = {'a': 10, 'b': 20}
d2 = {'a': 15, 'c': 5}
merged=d1.copy()
for k,v in d2.items():
    if k in merged:
        merged[k]=max(merged[k],v)
    else:
        merged[k]=v
print(merged)
```

```python
'''Question 15: Remove Duplicates from List of Dictionaries

Problem:
You have a list of dictionaries. Remove duplicates so that each dictionary appears only once.

Example:

Input: [{"a":1}, {"a":1}, {"b":2}]
Output: [{"a":1}, {"b":2}]'''

l=[{"a":1}, {"a":1}, {"b":2}]
lnew=[]
for i in range(len(l)):
    if l[i] not in l[:i]:
        lnew.append(l[i])
print(lnew)
```

```python
'''Question 17: Character Mapping

Problem:
You are given two strings of equal length. Create a dictionary that maps characters from the first string to characters in the second string.

Example:

Input: "abc", "123"
Output: {'a': '1', 'b': '2', 'c': '3'}'''

s1="abc"
s2="123"
mapping={}
for k,v in zip(s1,s2):
    mapping[k]=v
print(mapping)
```

```python
'''Question 18: Two Dictionaries Intersection

Problem:
Given two dictionaries, find common keys and return a new dictionary where the values are the sum of the values from both dictionaries.

Example:

d1 = {'a': 1, 'b': 2}
d2 = {'b': 3, 'c': 4}

Output: {'b': 5}'''

d1 = {'a': 1, 'b': 2}
d2 = {'b': 3, 'c': 4}
intersection={}
for k in d1:
    if k in d2:
        intersection[k]=d1[k]+d2[k]
print(intersection)
```

```python
'''Question 19: Remove Keys with Empty Values

Problem:
Given a dictionary, remove keys that have empty strings '' or None as values.

Example:

Input: {'a': '', 'b': 'Hello', 'c': None}
Output: {'b': 'Hello'}'''

d={'a': '', 'b': 'Hello', 'c': None}
for k,v in list(d.items()):
    if v=='' or v==None:
        d.pop(k)
print(d)
```
