```python
# Frequency Counting
# Count how many times each element appears in a list.

nums = [1, 2, 2, 3, 3, 3, 4]
freq={}

for i in nums:
    freq[i]=freq.get(i,0)+1

print(freq)
```

```python
# Count Only Repeated Elements

nums = [4, 2, 4, 3, 2, 4, 5]
freq={}

for i in nums:
    freq[i]=freq.get(i,0)+1

repeated={}

for k,v in freq.items():
    if v>1:
        repeated[k]=v

print(repeated)
```

```python
# First Non-Repeating Element Using Dictionary

s = "aabbcde"

freq={}

for i in s:
    freq[i]=freq.get(i,0)+1

print(freq)

for k,v in freq.items():
    if v==1:
        print(k)
        break
```

```python
# First Non-Repeating Number

nums = [4, 5, 4, 6, 5, 7, 8]

freq={}

for i in nums:
    freq[i]=freq.get(i,0)+1

for i in nums:
    if freq[i]==1:
        print(i)
        break
```

```python
# Find Duplicate Elements Using Dictionary

nums = [4, 2, 4, 3, 2, 4, 5]

freq={}

for i in nums:
    freq[i]=freq.get(i,0)+1

result=[]

for k,v in freq.items():
    if v>1:
        result.append(k)

print(result)
```

```python
# Two Sum Using Dictionary
# Given a list of numbers and a target, find the indices of two numbers whose sum equals the target.
# when you need not to sort the list

nums = [3, 2, 4]
target = 6

seen={}

for i,num in enumerate(nums):
    complement=target-num
    if complement in seen:
        print([seen[complement],i])
        break
    seen[num]=i
```

```python
# Check Anagrams Using Dictionary

s1 = "aabbcc"
s2 = "abcabc"

d1={}
d2={}

if len(s1) != len(s2):
    print(False)
else:
    for i in s1:
        d1[i]=d1.get(i,0)+1

    for i in s2:
        d2[i]=d2.get(i,0)+1

print(d1==d2)
```

```python
# Given a list of strings, group the words that are anagrams of each other.

words = ["eat", "tea", "tan", "ate", "nat", "bat"]

dict={}

for i in words:
    sort_i=''.join(sorted(i))
    if sort_i not in dict:
        dict[sort_i]=[]
    dict[sort_i].append(i)

print(dict)

print(list(dict.values()))
```

```python
# Find Most and Least Frequent Elements Using Dictionary

nums = [4, 2, 4, 3, 2, 4, 5]

freq={}

for i in nums:
    freq[i]=freq.get(i,0)+1

print(freq)
print(max(freq, key=freq.get))
print(min(freq, key=freq.get))
```

```python
nums = [1, 2, 2, 3, 3, 4, 4, 5]

freq = {}

for i in nums:
    freq[i] = freq.get(i, 0) + 1

print(freq)

max_value=max(freq.values())
print(max_value)

min_value=min(freq.values())
print(min_value)

max_res=[]
min_res=[]

for k,v in freq.items():
    if v==max_value:
        max_res.append(k)
    if v==min_value:
        min_res.append(k)

print(max_res)
print(min_res)
```

```python
# Group Numbers as Even and Odd

nums = [1, 2, 3, 4, 5, 6, 7, 8]

dict={}

for i in nums:
    if i%2==0:
        if 'even' not in dict:
            dict['even']=[]
        dict['even'].append(i)

    else:
        if 'odd' not in dict:
            dict['odd']=[]
        dict['odd'].append(i)

print(dict)
        
```

```python
# Group Words by Their Length

words = ["cat", "dog", "apple", "bat", "banana"]

dict={}

for i in words:
    if len(i) not in dict:
        dict[len(i)]=[]
    dict[len(i)].append(i)

print(dict)
```
