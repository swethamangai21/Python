```python
# Linear Traversal- Linear traversal means visiting every element of a list one by one to find something

'''
-Largest or smallest element
-Second largest element
-Sum of elements
-Count of even or odd numbers
-Elements satisfying a conditio
'''
```

```python
# Find the Largest Element

nums = [10, 5, 20, 8, 15]

largest=nums[0]

for i in nums:
    if i>largest:
        largest=i

print(largest)
```

```python
# Find the Smallest Element

nums = [12, 4, 7, 1, 19, 3]

smallest=nums[0]

for i in nums:
    if i<smallest:
        smallest=i

print(smallest)
```

```python
# Find the Second Largest Element

nums = [20, 10, 5, 8]

first_largest=float('-inf')
second_largest=float('-inf')

for i in nums:
    if i>first_largest:
        second_largest=first_largest
        first_largest=i
    elif i < first_largest and i > second_largest:
        second_largest=i

print(second_largest)
```

```python
# Hash Set- Python data structure that stores unique elements

'''
-Find duplicates
-Check whether duplicates exist
-Find unique elements
-Remove duplicates 
-Check whether an element has appeared before
'''
```

```python
# Find the First Duplicate Element

# Given a list, find the first element that appears for the second time while traversing from left to right.

nums = [4, 2, 7, 2, 4, 9]

# step 1: create an empty set
seen=set()

# step 2: traverse the list
for i in nums:
    if i in seen: # step 3: check whether the number already exists
        print(i)
        break
    else:
        seen.add(i)
```

```python
# Check Whether a List Contains Duplicates

nums = [1, 2, 3, 4, 2]

seen=set()

isDuplicate=False

for i in nums:
    if i in seen:
        isDuplicate=True
        break
    else:
        seen.add(i)

print(isDuplicate)
```

```python
# Remove Duplicates Without Changing Order

nums = [4, 2, 4, 3, 2, 1]

seen=[]

for i in nums:
    if i not in seen:
        seen.append(i)

print(seen)
```

```python
# Hash Map (Dictionary)- we use a dictionary to store data in key-value pairs

'''
-Count occurrences
-Frequency of elements
-Most frequent element
-First non-repeating element
-Find pairs with a target sum
'''
```

```python
# Count the Frequency of Each Element

nums = [1, 2, 2, 3, 3, 3]

# step 1: create an empty dictionary
freq={}

# step 2: traverse the list
for i in nums:
    freq[i]=freq.get(i,0)+1 # step 3: count each element

print(freq)

'''
nums = [1, 2, 2, 3, 3, 3]

freq={}

for i in nums:
    if i in freq:
        freq[i]+=1
    else:
        freq[i]=1
'''
```

```python
# Find the Most Frequent Element

nums = [1, 2, 2, 3, 3, 3, 4]

freq={}

for i in nums:
    freq[i]=freq.get(i, 0)+1

print(freq)

max_count=0
result=None

for k, v in freq.items():
    if v > max_count:
        max_count=v
        result=k

print(result)

# another way: max(freq.values())
```

```python
# Find Elements Appearing Only Once

nums = [1, 2, 2, 3, 4, 4, 5]

freq={}

for i in nums:
    freq[i]=freq.get(i, 0)+1

print(freq)

result=[]

for k, v in freq.items():
    if v==1:
        result.append(k)

print(result)
```

```python
# Two Pointers- Two Pointers means using two positions (indices) to solve a list problem efficiently.

'''
1) Opposite-direction pointers: 
One starts at the beginning, another at the end- Two Sum (sorted), reverse list

2) Same-direction pointers:
Both start at the beginning and move forward- Move zeros, remove duplicates from sorted list
'''
```

```python
# Move All Zeros to the End

# i(right): Checks every number to find non-zero elements
# left: Keeps track of where the next non-zero number should go

nums = [0, 1, 0, 3, 12]

# step 1: initialize the pointer
left=0

# step 2: traverse
for i in range(len(nums)):
    if nums[i] != 0: # step 3: check condition
        nums[left], nums[i] = nums[i], nums[left] # step 4: swap
        left+=1 # step 5: move the pointer

print(nums)
```

```python
# Move All Ones to the Beginning

nums = [0, 1, 0, 1, 1, 0]

left=0

for i in range(len(nums)):
    if nums[i] == 1:
        nums[left], nums[i] = nums[i], nums[left]
        left+=1

print(nums)
```

```python
# Sorting

'''
-Find the missing number
-Merge two sorted lists
-Find consecutive numbers
-Find duplicates 
-Find the kth largest or smallest element
'''
```

```python
# Find the Missing Number

nums = [1, 4, 2, 5]

# step 1: sort the list
nums.sort()

# step 2: traverse the list
for i in range(len(nums)-1): # to compare each element with next element
    # step 3: check whether consecutive numbers differ by 1 eg: 1 → 2  Difference = 1, 2 → 3  Difference = 1
    if nums[i+1]-nums[i]>1:
        missing=nums[i]+1 # step 4: find the missing number
        break

print(missing)
```

```python
# Find Duplicate Elements Using Sorting

nums = [4, 2, 1, 4, 3, 2, 5]

nums.sort()

for i in range(len(nums)-1):
    if nums[i]==nums[i+1]:
        print(nums[i])
```

```python
# prefix sum- calculating the running sum of elements from the beginning of a list

'''
step 1: initialize total=0 --> store the running sum
step 2: create an empty list prefix=[] --> store the cumulatice sums
step 3: traverse
step 4: add the current number to total
step 5: append total to prefix
'''
'''
-Running sum
-Cumulative sum
-Sum between two indiced
-Range sum queries
-Subarray sum
'''
```

```python
# Find the Running Sum of a List

nums = [2, 4, 6, 8, 10]

total=0
prefix=[]

for i in nums:
    total+=i
    prefix.append(total)

print(prefix)
```

```python
# Find Sum Between Two Indices

nums = [3, 5, 2, 8, 6]

left = 1
right = 3

total=0
prefix=[]

for i in nums:
    total+=i
    prefix.append(total)


if left==0:
    result=prefix[right]
else:
    result=prefix[right]-prefix[left-1]

print(nums)
print(prefix)
print(result)
```
