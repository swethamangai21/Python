```python
'''Question 11: Most Frequent Element Replacement (Set-based)

Problem:
Given a tuple:

t = (3, 5, 2, 3, 7, 2, 3)

Find the most frequent element using a set.

Replace all occurrences of it with 0 in a new tuple.

Print the most frequent element and the new tuple.'''

t = (3, 5, 2, 3, 7, 2, 3)
tcount=0
for i in range(len(t)):
    if t[i] not in t[:i]:
        if t.count(t[i])>tcount:
            tcount=t.count(t[i])
            freq=t[i]
tlist=list(t)
for i in range(len(tlist)):
    if tlist[i]==freq:
        tlist[i]=0
print(tlist)
```

```python
'''Question 13: Flatten Nested List with Unique Elements

Problem:

Given a nested list:

nested = [1, 2, [2, 3], [3, 4, [4, 5]]]

Flatten all integers into a single set of unique elements.

Print the final set.'''

nested = [1, 2, [2, 3], [3, 4, [4, 5]]]
flat=[]
for i in range(len(nested)):
    if isinstance(nested[i],list):
        for j in range(len(nested[i])):
            if isinstance(nested[i][j],list):
                for k in range(len(nested[i][j])):
                    flat.append(nested[i][j][k])
            else:
                flat.append(nested[i][j])
    else:
        flat.append(nested[i])

print(set(flat))
```
