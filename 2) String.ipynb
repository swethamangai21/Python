```python
'''1. Reverse a String

Write a program to reverse a given string.

Example
Input: "python"
Output: "nohtyp"'''

s='Python'
print(s[::-1])
```

```python
'''2. Check Palindrome

Check whether a string is a palindrome.

Example
Input: "madam"
Output: Palindrome'''

s='madam'
if s[:]==s[::-1]:
    print('Palindrome')
else:
    print('Not a Palindrome')
```

```python
'''3. Count Vowels and Consonants

Count the number of vowels and consonants in a string.

Example
Input: "python"
Output:
Vowels: 1
Consonants: 5'''

s='Python'
count_v=0
count_c=0
for i in s:
    if i=='a' or i=='e' or i=='i' or i=='o' or i=='u' or i=='A' or i=='E' or i=='I' or i=='O' or i=='U':
        count_v+=1
    else:
        count_c+=1
print("Vowels:",count_v)
print("Consonants:",count_c)

'''s = "Python"
count_v = 0
count_c = 0

for i in s:
    if i.isalpha():                     # ensures only letters are counted
        if i.lower() in "aeiou":        # checks vowel
            count_v += 1
        else:
            count_c += 1

print("Vowels:", count_v)
print("Consonants:", count_c)'''
```

```python
'''4. Count Frequency of Each Character

Print the frequency of each character in a string.

Example
Input: "banana"
Output

b:1
a:3
n:2'''

s='banana'
for i in range(len(s)):
    if s[i] not in s[:i]:
        print(s[i],':',s.count(s[i]))
```

```python
'''7. Check Anagram

Check if two strings are anagrams.

Example

listen
silent

Output: Anagram'''

s1='listen'
s2='silent'

if sorted(s1)==sorted(s2):
    print('Anagram')
else:
    print('Not an Anagram')
```

```python
'''8. Find Duplicate Characters

Print duplicate characters in a string.

Example
Input: "programming"
Output: r g m'''

s='programming'
for i in range(len(s)):
    if s[i] not in s[:i]:
        if s.count(s[i])>1:
            print(s[i])
```

```python
'''9. Remove Duplicate Characters

Remove duplicate characters but keep first occurrence.

Example
Input: "programming"
Output: "progamin" '''

s='programmingi'
snew=''
for i in range(len(s)):
    if s[i] not in s[:i]:
        snew+=s[i]
print(snew)
```

```python
'''10. Longest Word in a Sentence

Find the longest word in a sentence.

Example
Input
"Python is a powerful programming language"

Output
programming'''

s="Python is a powerful programming language"
smax=''
for i in s.split(' '):
    if len(i)>len(smax):
        smax=i
print(smax)
```

```python
'''13. Capitalize First Letter of Every Word (without title())

Example

Input
"python is powerful"

Output
"Python Is Powerful" '''

s="python is powerful"
snew=''
for i in s.split(' '):
    snew+=i[0].upper()+i[1:]+' '
print(snew)
```

```python
'''15. Find All Substrings of a String

Example

Input "abc"

Output

a
ab
abc
b
bc
c'''

s = "abc"
for i in range(len(s)):
    for j in range(i+1, len(s)+1):
        print(s[i:j])
```

```python
'''16. String Compression

Compress repeated characters.

Example

Input
"aaabbccdaa"

Output
a3b2c2d1a2'''

s="aaabbccdaa"
scount=1
for i in range(1,len(s)):
    if s[i]==s[i-1]:
        scount+=1
    else:
        print(s[i-1]+str(scount),end="")
        scount=1
print(s[-1]+str(scount))
```

```python
'''17. Longest Substring Without Repeating Characters

Example

Input
"abcabcbb"

Output
3

Substring: "abc" '''

s="abcabcbb"
stemp=''
maxlen=0

for i in range(len(s)):
    
    if s[i] in stemp:
        stemp = stemp[stemp.index(s[i])+1:]
    
    stemp += s[i]
    
    if len(stemp) > maxlen:
        maxlen = len(stemp)

print(maxlen)
```

```python
'''18. Check Balanced Parentheses

Example

Input: "((()))"
Output: Balanced
Input: "(()"
Output: Not Balanced'''

s="((()))"
count=0

for i in s:
    
    if i=="(":
        count+=1
    else:
        count-=1
    
    if count<0:
        print("Not Balanced")
        break

if count==0:
    print("Balanced")
else:
    print("Not Balanced")
    
```

```python
'''19. Find the Most Frequent Character

Example

Input
"success"

Output
s'''

s="success"
smax=0
schar=''
for i in range(len(s)):
    if s[i] not in s[:i]:
        if s.count(s[i])>smax:
            smax=s.count(s[i])
            schar=s[i]
print(schar)
```

```python
'''20. Group Anagrams from List of Words

Example

Input

["eat","tea","tan","ate","nat","bat"]

Output

[["eat","tea","ate"],
 ["tan","nat"],
 ["bat"]]'''

s = ["eat","tea","tan","ate","nat","bat"]
d = {}

for i in s:
    k = ''.join(sorted(i))
    
    if k not in d:
        d[k] = []
        
    d[k].append(i)

print(d.values())
```
